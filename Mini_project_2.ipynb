```python
!pip install -q ultralytics gradio
from ultralytics import YOLO
import gradio as gr
from PIL import Image
model = YOLO("yolov8n.pt")  # Note: Changed 'yolo26n.pt' to 'yolov8n.pt' as yolo26 is not a standard release
VEHICLE_CLASSES = {
    2: "Car",
    3: "Motorcycle",
    5: "Bus",
    7: "Truck"
}
def detect_vehicles(image, confidence):
    if image is None:
        return None, "Please upload an image."
    results = model.predict(source=image, conf=confidence, imgsz=640, verbose=False)
    result = results[0]
    output_image = result.plot()
    output_image = Image.fromarray(output_image[:, :, ::-1])
    vehicle_list = []
    if result.boxes is not None:
        for box in result.boxes:
            class_id = int(box.cls[0])
            conf = float(box.conf[0])
            if class_id in VEHICLE_CLASSES:
                vehicle_name = VEHICLE_CLASSES[class_id]
                vehicle_list.append(f"{vehicle_name} - {conf*100:.1f}%")
    if vehicle_list:
        message = "Vehicles Detected:\n"
        message += "\n".join(vehicle_list)
        message += f"\n\nTotal Vehicles: {len(vehicle_list)}"
    else:
        message = "No vehicles detected."
    return output_image, message
app = gr.Interface(
    fn=detect_vehicles,
    inputs=[
        gr.Image(type="pil", label="Upload Vehicle Image"),
        gr.Slider(minimum=0.10, maximum=0.90, value=0.25, step=0.05, label="Detection Confidence")
    ],
    outputs=[
        gr.Image(type="pil", label="Detected Vehicles"),
        gr.Textbox(label="Detection Result")
    ],
    title="Vehicle Detection Using CNN",
    description=(
        "Upload a road or vehicle image. "
        "The CNN-based YOLO model detects cars, "
        "motorcycles, buses and trucks."
    ),
    examples=[]
)
app.launch(share=True)
```
